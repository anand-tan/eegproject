```python
import mne

edf_files = [
    "chb-mit-scalp-eeg-database-1.0.0/chb01/chb01_01.edf",
    "chb-mit-scalp-eeg-database-1.0.0/chb01/chb01_03.edf",
]

for path in edf_files:
    print(f"Loading {path}")
    raw = mne.io.read_raw_edf(path, preload=True, verbose=False)

    # Clean channel names
    raw.rename_channels(lambda x: x.strip("."))

    # Ensure 256 Hz
    if raw.info["sfreq"] != 256:
        raw.resample(256)

    # Basic filtering (optional)
    raw.filter(1.0, 40.0, fir_design="firwin")
    raw.notch_filter(freqs=[60])

    # INTERACTIVE EEG PLOT
    raw.plot(
        duration=10,     # seconds visible at once
        n_channels=18,   # how many channels shown
        scalings="auto",
        title=path
    )
```
